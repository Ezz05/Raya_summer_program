# CNN From Scratch on MNIST (NumPy Only)

This notebook implements a simple feedforward neural network **from scratch using only NumPy**
(no TensorFlow/PyTorch layers, autograd, or built-in optimizers) to classify handwritten digits
from the MNIST dataset.

**Goals:**
- Implement forward propagation, the loss function, and backpropagation manually.
- Verify the manual backpropagation gradients using numerical gradient checking.
- Train the network with full-batch gradient descent and evaluate its accuracy.

**Architecture:**
- Input layer: 784 neurons (28×28 flattened, normalized pixel values)
- Hidden layer: 128 neurons, ReLU activation
- Output layer: 10 neurons, softmax activation (one-hot digit classification)


## 1. Imports & Data Loading

```
import numpy as np
from tensorflow.keras.datasets import mnist

# Load the MNIST dataset (60,000 training images, 10,000 test images, 28x28 grayscale)
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

## 2. Preprocessing

```
# Flatten each 28x28 image into a 784-length vector
x_train_flat = x_train.reshape(60000, 784)
x_test_flat = x_test.reshape(10000, 784)

# Normalize pixel values from [0, 255] to [0, 1] for stable training
x_train_norm = x_train_flat / 255
x_test_norm = x_test_flat / 255

# One-hot encode the integer labels (0-9) into 10-dimensional vectors
identity = np.eye(10)
y_train_encoded = identity[y_train]
y_test_encoded = identity[y_test]

# Transpose so that each COLUMN is one example: shape (features, num_examples)
# This layout matches the matrix-multiplication convention used throughout (W @ X + b)
x_train_final = x_train_norm.T
x_test_final = x_test_norm.T
y_train_final = y_train_encoded.T
y_test_final = y_test_encoded.T
```

## 3. Parameter Initialization

```
# Fix the random seed for reproducibility
np.random.seed(42)

# Weight matrices are initialized with small random values to break symmetry;
# biases are initialized to zero (standard practice)
W1 = np.random.randn(128, 784) * 0.01   # hidden layer weights: (hidden_units, input_features)
b1 = np.zeros((128, 1))                 # hidden layer biases
W2 = np.random.randn(10, 128) * 0.01    # output layer weights: (num_classes, hidden_units)
b2 = np.zeros((10, 1))                  # output layer biases
```

## 4. Activation Functions

```
def relu(Z):
    """ReLU activation: element-wise max(0, Z). Used in the hidden layer."""
    return np.maximum(Z, 0)


def softmax(Z):
    """
    Softmax activation for the output layer, converting raw scores (logits)
    into a probability distribution over the 10 digit classes.

    Subtracting the column-wise max before exponentiating (Z_shifted) is a
    standard numerical-stability trick that prevents overflow, without
    changing the resulting probabilities.
    """
    Z_shifted = Z - np.max(Z, axis=0, keepdims=True)
    exp_Z = np.exp(Z_shifted)
    return exp_Z / np.sum(exp_Z, axis=0, keepdims=True)


# Quick sanity check: softmax output should sum to 1
test_Z = np.array([[2.0], [1.0], [0.1]])
print(softmax(test_Z))
```

## 5. Forward Propagation

```
def forward_prop(x, w1, w2, b1, b2):
    """
    Runs one forward pass through the network.

    Layer 1 (hidden):  Z1 = W1 . x + b1  ->  A1 = ReLU(Z1)
    Layer 2 (output):  Z2 = W2 . A1 + b2 ->  A2 = softmax(Z2)

    Returns:
        A2: predicted class probabilities, shape (10, num_examples)
        cache: intermediate values (Z1, A1, Z2, A2) needed for backpropagation
    """
    Z1 = w1 @ x + b1
    A1 = relu(Z1)
    Z2 = w2 @ A1 + b2
    A2 = softmax(Z2)
    cache = (Z1, A1, Z2, A2)
    return A2, cache
```

## 6. Loss Function

```
def compute_loss(A2, Y):
    """
    Categorical cross-entropy loss, averaged over all examples in the batch.

    Y is one-hot encoded, so Y * log(A2) picks out the log-probability
    the model assigned to the correct class for each example.
    """
    m = Y.shape[1]
    loss = -1 / m * np.sum(Y * np.log(A2))
    return loss
```

## 7. Backward Propagation

```
def backward_propagation(X, Y, W1, b1, W2, b2, cache):
    """
    Computes gradients of the loss with respect to every parameter
    (W1, b1, W2, b2) via manual backpropagation (chain rule).

    Key derivative used: for softmax + cross-entropy combined, dZ2 = A2 - Y
    (a well-known simplification that avoids computing the softmax Jacobian
    directly).
    """
    Z1, A1, Z2, A2 = cache
    m = Y.shape[1]

    # Output layer gradients
    dz2 = A2 - Y
    dW2 = (1 / m) * dz2 @ A1.T
    db2 = (1 / m) * np.sum(dz2, axis=1, keepdims=True)

    # Hidden layer gradients (backprop through ReLU: derivative is 1 where Z1 > 0, else 0)
    dA1 = W2.T @ dz2
    dZ1 = dA1 * (Z1 > 0)
    dW1 = (1 / m) * dZ1 @ X.T
    db1 = (1 / m) * np.sum(dZ1, axis=1, keepdims=True)

    grads = (dW1, db1, dW2, db2)
    return grads
```

## 8. Parameter Update (Gradient Descent Step)

```
def update_parameters(W1, b1, W2, b2, dW1, db1, dW2, db2, learning_rate):
    """Standard gradient descent update: param = param - learning_rate * gradient."""
    W1 = W1 - learning_rate * dW1
    b1 = b1 - learning_rate * db1
    W2 = W2 - learning_rate * dW2
    b2 = b2 - learning_rate * db2
    return W1, b1, W2, b2
```

## 9. Sanity Check: One Forward + Backward + Update Pass\n\nBefore training for real, run a single pass end-to-end to confirm shapes match and the loss decreases after one update step.

```
# One forward pass on the full training set (untrained network)
A2, cache = forward_prop(x_train_final, W1, W2, b1, b2)
loss = compute_loss(A2, y_train_final)
dW1, db1, dW2, db2 = backward_propagation(x_train_final, y_train_final, W1, b1, W2, b2, cache)

print("Loss:", loss)                          # expect ~2.3 (untrained network, ~ln(10))
print("dW1:", dW1.shape, "  W1:", W1.shape)    # gradient shape should match parameter shape
print("db1:", db1.shape, "  b1:", b1.shape)
print("dW2:", dW2.shape, "  W2:", W2.shape)
print("db2:", db2.shape, "  b2:", b2.shape)

# Apply a single gradient descent update
learning_rate = 0.1
W1, b1, W2, b2 = update_parameters(W1, b1, W2, b2, dW1, db1, dW2, db2, learning_rate)

# Recompute the loss to confirm it decreased even slightly
A2, cache = forward_prop(x_train_final, W1, W2, b1, b2)
new_loss = compute_loss(A2, y_train_final)
print("New loss after one update:", new_loss)
```

## 10. Gradient Checking\n\nTo verify the manual backpropagation math is correct, compare the analytical gradients from `backward_propagation` against numerical gradients estimated via finite differences: `(loss(param + eps) - loss(param - eps)) / (2 * eps)`. If backprop is implemented correctly, these should match closely (relative error < 1e-5).

```
def gradient_check(param, dparam, param_name, X, Y, W1, b1, W2, b2, num_checks=10, epsilon=1e-7):
    """
    Numerically verifies a subset of gradients for a given parameter by
    perturbing individual entries and comparing the resulting finite-difference
    estimate against the analytical gradient from backpropagation.
    """
    print(f"Checking {param_name} ({num_checks} random entries):")
    flat_param = param.reshape(-1)
    flat_dparam = dparam.reshape(-1)

    for _ in range(num_checks):
        idx = np.random.randint(0, flat_param.size)
        original_value = flat_param[idx]

        # Nudge the parameter up and recompute the loss
        flat_param[idx] = original_value + epsilon
        A2_plus, _ = forward_prop(X, W1, W2, b1, b2)
        loss_plus = compute_loss(A2_plus, Y)

        # Nudge the parameter down and recompute the loss
        flat_param[idx] = original_value - epsilon
        A2_minus, _ = forward_prop(X, W1, W2, b1, b2)
        loss_minus = compute_loss(A2_minus, Y)

        # Restore the original value before moving to the next check
        flat_param[idx] = original_value

        grad_approx = (loss_plus - loss_minus) / (2 * epsilon)
        grad_backprop = flat_dparam[idx]

        rel_error = abs(grad_approx - grad_backprop) / (abs(grad_approx) + abs(grad_backprop) + 1e-10)

        print(f"  idx={idx:4d}  approx={grad_approx: .8f}  backprop={grad_backprop: .8f}  rel_error={rel_error:.2e}")


# Use a small slice of real MNIST data for the check (gradient checking is expensive,
# so we don't run it on the full 60,000-example training set)
X_check = x_train_final[:, :5]
Y_check = y_train_final[:, :5]

# Run one forward + backward pass to get the gradients we want to verify
A2_check, cache_check = forward_prop(X_check, W1, W2, b1, b2)
dW1_check, db1_check, dW2_check, db2_check = backward_propagation(X_check, Y_check, W1, b1, W2, b2, cache_check)

# Check each parameter
gradient_check(W1, dW1_check, "W1", X_check, Y_check, W1, b1, W2, b2)
gradient_check(b1, db1_check, "b1", X_check, Y_check, W1, b1, W2, b2)
gradient_check(W2, dW2_check, "W2", X_check, Y_check, W1, b1, W2, b2)
gradient_check(b2, db2_check, "b2", X_check, Y_check, W1, b1, W2, b2)
```

## 11. Training Loop

```
def train(X, Y, X_test, Y_test, n_h, learning_rate, num_iterations):
    """
    Trains a 2-layer neural network from scratch using full-batch gradient descent.

    Args:
        X, Y: training data and one-hot labels, shape (features, m) / (classes, m)
        X_test, Y_test: held-out test data/labels, used only for monitoring
        n_h: number of hidden units
        learning_rate: gradient descent step size
        num_iterations: number of full-batch gradient descent steps

    Returns:
        Trained parameters (W1, b1, W2, b2) plus history lists (recorded every
        10 iterations) for loss, training accuracy, and test accuracy.
    """
    n_x = X.shape[0]
    n_y = Y.shape[0]

    # Re-initialize parameters (fresh training run)
    np.random.seed(42)
    W1 = np.random.randn(n_h, n_x) * 0.01
    b1 = np.zeros((n_h, 1))
    W2 = np.random.randn(n_y, n_h) * 0.01
    b2 = np.zeros((n_y, 1))

    loss_history = []
    train_acc_history = []
    test_acc_history = []

    for i in range(num_iterations):
        # Forward pass + loss on the full training set
        A2, cache = forward_prop(X, W1, W2, b1, b2)
        loss = compute_loss(A2, Y)

        # Backward pass + parameter update
        dW1, db1, dW2, db2 = backward_propagation(X, Y, W1, b1, W2, b2, cache)
        W1, b1, W2, b2 = update_parameters(W1, b1, W2, b2, dW1, db1, dW2, db2, learning_rate)

        # Log progress every 10 iterations
        if i % 10 == 0:
            train_preds = np.argmax(A2, axis=0)
            train_labels = np.argmax(Y, axis=0)
            train_acc = np.mean(train_preds == train_labels)

            A2_test, _ = forward_prop(X_test, W1, W2, b1, b2)
            test_preds = np.argmax(A2_test, axis=0)
            test_labels = np.argmax(Y_test, axis=0)
            test_acc = np.mean(test_preds == test_labels)

            loss_history.append(loss)
            train_acc_history.append(train_acc)
            test_acc_history.append(test_acc)

            print(f"Iteration {i}: loss={loss:.4f}, train_acc={train_acc:.4f}, test_acc={test_acc:.4f}")

    return W1, b1, W2, b2, loss_history, train_acc_history, test_acc_history


# Train the network: 128 hidden units, learning rate 0.5, 500 full-batch iterations
W1, b1, W2, b2, loss_history, train_acc_history, test_acc_history = train(
    x_train_final, y_train_final, x_test_final, y_test_final,
    n_h=128, learning_rate=0.5, num_iterations=500)
```

## 12. Final Evaluation

```
def predict(X, W1, b1, W2, b2):
    """Runs a forward pass and returns the predicted class (0-9) for each example."""
    A2, _ = forward_prop(X, W1, W2, b1, b2)
    predictions = np.argmax(A2, axis=0)
    return predictions


def compute_accuracy(predictions, Y):
    """Compares predicted classes against one-hot ground-truth labels Y."""
    true_labels = np.argmax(Y, axis=0)
    accuracy = np.mean(predictions == true_labels)
    return accuracy


train_predictions = predict(x_train_final, W1, b1, W2, b2)
test_predictions = predict(x_test_final, W1, b1, W2, b2)

train_accuracy = compute_accuracy(train_predictions, y_train_final)
test_accuracy = compute_accuracy(test_predictions, y_test_final)

print(f"Final training accuracy: {train_accuracy:.4f}")
print(f"Final test accuracy: {test_accuracy:.4f}")
```

## 13. Visualization

```
import matplotlib.pyplot as plt


def plot_loss_curve(loss_history):
    """Plots training loss recorded every 10 iterations."""
    iterations = [i * 10 for i in range(len(loss_history))]
    plt.figure(figsize=(8, 5))
    plt.plot(iterations, loss_history)
    plt.xlabel('Iteration')
    plt.ylabel('Loss')
    plt.title('Training Loss Over Time')
    plt.grid(True)
    plt.show()


def plot_accuracy_curve(train_acc_history, test_acc_history):
    """Plots training vs. test accuracy over the course of training."""
    iterations = [i * 10 for i in range(len(train_acc_history))]
    plt.figure(figsize=(8, 5))
    plt.plot(iterations, train_acc_history, label='Train Accuracy')
    plt.plot(iterations, test_acc_history, label='Test Accuracy')
    plt.xlabel('Iteration')
    plt.ylabel('Accuracy')
    plt.title('Accuracy Over Time')
    plt.legend()
    plt.grid(True)
    plt.show()


def plot_sample_predictions(X_test_original, X_test_final, y_test, W1, b1, W2, b2, num_samples=9):
    """Displays a grid of random test images with predicted vs. true labels
    (green title = correct prediction, red = incorrect)."""
    indices = np.random.choice(X_test_original.shape[0], num_samples, replace=False)

    plt.figure(figsize=(9, 9))
    for i, idx in enumerate(indices):
        image = X_test_original[idx]
        true_label = y_test[idx]

        single_input = X_test_final[:, idx].reshape(784, 1)
        predicted = predict(single_input, W1, b1, W2, b2)[0]

        plt.subplot(3, 3, i + 1)
        plt.imshow(image, cmap='gray')
        color = 'green' if predicted == true_label else 'red'
        plt.title(f"True: {true_label}, Pred: {predicted}", color=color)
        plt.axis('off')

    plt.tight_layout()
    plt.show()


plot_loss_curve(loss_history)
plot_accuracy_curve(train_acc_history, test_acc_history)
plot_sample_predictions(x_test, x_test_final, y_test, W1, b1, W2, b2)
```

## Results & Write-up

**Architecture:** A 2-layer neural network (1 hidden layer) built from scratch using only NumPy:
- Input layer: 784 neurons (28×28 flattened, normalized pixel values)
- Hidden layer: 128 neurons, ReLU activation
- Output layer: 10 neurons, softmax activation

**Training setup:**
- Loss function: cross-entropy
- Optimization: full-batch gradient descent
- Learning rate: 0.5
- Iterations: 500
- Random seed fixed at 42 for reproducibility

**Results:**
- Final training accuracy: 95.53%
- Final test accuracy: 95.29%

**Gradient checking:** Backpropagation was verified against numerical gradients (finite differences) on a small batch of real MNIST data. Relative errors for all tested parameters (W1, b1, W2, b2) were on the order of 1e-6 to 1e-10, well within the acceptable threshold (<1e-5), confirming the backprop implementation is mathematically correct.

**Observations:** The training and test accuracies are very close (95.53% vs 95.29%), indicating the model generalizes well and is not overfitting.

**Conclusion:** The network exceeded a 90% test accuracy target by a healthy margin using basic full-batch gradient descent, without momentum, regularization, or mini-batching — demonstrating that a simple 2-layer network is sufficient for the MNIST classification task.
